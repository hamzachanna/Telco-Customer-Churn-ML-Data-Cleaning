## Part 4 — Data Cleaning Decisions

Cleaning is a series of justified decisions, not a fixed recipe. For every issue you fix, you must be able to explain why you chose that method over an alternative.

*Concepts/functions you may find useful:* `pd.to_numeric(s, errors='coerce')`, `df.fillna(value)`, `df.dropna()`, `df.drop_duplicates()`, `df[col].str.strip()`, `df.astype(dtype)`

**Tasks**
10. For every data-quality issue you found in Part 3, record: Problem → Decision → Method Used → Reason.
11. Apply your cleaning decisions to produce an intermediate cleaned dataframe, and confirm (with code output) that each issue is actually resolved.

**Task 10 — Cleaning Decision Log**

| Problem | Decision | Method Used | Reason |
|---|---|---|---|
| _TotalCharges stored as text_ | _convert those into a null val, then reallocate 0 to those_ | _pd.toNumeric() on a condition where it has a blank space_ | _as i mentioned earlier: to achieve meaningfull dataset_ |
| _TotalCharges type mismatch_ | _typecasting from string to float_ | _.astype(float): lets you typecast_ | _for EDA numeric datatypes are required , for that reason i had to do it with float_ |

```python
# Task 11 — apply your cleaning decisions here
df_clean = df.copy()

for col in df_clean.select_dtypes(include=['object']).columns:
    df_clean[col] = df_clean[col].astype(str).str.strip()

df_clean['TotalCharges'] = pd.to_numeric(df_clean['TotalCharges'], errors='coerce').fillna(0.0)

df_clean = df_clean.drop_duplicates(subset=['customerID'], keep='first')

print("Data cleaning verification:")
print(f"TotalCharges type: {df_clean['TotalCharges'].dtype}")
print(f"TotalCharges null count: {df_clean['TotalCharges'].isnull().sum()}")

zero_tenure_charges = df_clean[df_clean['tenure'] == 0]['TotalCharges'].tolist()
print(f"Charges for tenure = 0: {zero_tenure_charges}")

print(f"Duplicate rows: {df_clean.duplicated().sum()}")
print(f"Duplicate IDs: {df_clean['customerID'].duplicated().sum()}")
print(f"Total dataset nulls: {df_clean.isnull().sum().sum()}")
```

```python
# Task 11 (continued) — confirm each issue is resolved (e.g. re-check isnull().sum(), duplicated().sum(), dtypes)

print(df_clean[['tenure', 'MonthlyCharges', 'TotalCharges']].dtypes)

print(df_clean.isnull().sum())

print(f"Full duplicate rows: {df_clean.duplicated().sum()}")
print(f"Duplicate customerIDs: {df_clean['customerID'].duplicated().sum()}")
print(df_clean[df_clean['tenure'] == 0][['tenure', 'MonthlyCharges', 'TotalCharges']])
```

## Part 7 — Bivariate Analysis: Features vs. Churn

Now connect individual features to the target. You choose the appropriate visualization or summary for each feature type — a categorical feature is not explored the same way as a numerical one.

*Concepts/functions you may find useful:* `pd.crosstab(df[col], df['Churn'], normalize='index')`, `df.groupby(col)['Churn']...`, `sns.boxplot(x='Churn', y=col, data=df)`

**Tasks**
16. Investigate at least 2 categorical features (e.g. Contract, InternetService, PaymentMethod) against Churn. For each, record: Observation / Evidence / Possible Explanation.
17. Investigate at least 2 numerical features (e.g. tenure, MonthlyCharges) against Churn using the same Observation / Evidence / Possible Explanation format.

```python
# Task 16 — categorical feature 1 vs Churn
cat_cols = ["Contract", "InternetService", "PaymentMethod"]

for col in cat_cols:
    print(f"=== Normalized Cross-Tabulation: {col} vs Churn ===")

    ct = pd.crosstab(df[col], df["Churn"], normalize="index") * 100
    print(ct.round(2))
    print("\n")

    ct.plot(kind="bar", stacked=True, figsize=(8, 4), colormap="coolwarm")
    plt.title(f"Churn Rate by {col}")
    plt.ylabel("Percentage (%)")
    plt.xlabel(col)
    plt.legend(title="Churn", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.xticks(rotation=45 if col == "PaymentMethod" else 0)
    plt.tight_layout()
    plt.show()
```

**Observation:** _It has been observed that customers who use electronic check, fiber optic and month-month contract have highest churn rate._
**Evidence:** _the evidence is shown in graphs_
**Possible Explanation:** _no by applying intersection among these 3 attributes we can recognize the pattern in customers_

```python
# Task 16 — categorical feature 2 vs Churn
internet_ct = pd.crosstab(df["InternetService"], df["Churn"], normalize="index") * 100
print(internet_ct)

plt.figure(figsize=(7, 4))
sns.barplot(x=internet_ct.index, y=internet_ct["Yes"])
plt.title("Churn Percentage by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.ylim(0, 100)
plt.show()
```

**Observation:** _It has been observed that approx 40% of churn rate is of customers who are using fiber optic internet_
**Evidence:** _Clearly mentioned on y-axis of the graph above_
**Possible Explanation:** _Maybe because it is already too costly for customers and the company doesnt offer any discounts for customers using fiber optics._

```python
# Task 17 — numerical feature 1 vs Churn
tenure_stats = df.groupby("Churn")["tenure"].describe()
print(tenure_stats)

plt.figure(figsize=(7, 4))
sns.boxplot(x="Churn", y="tenure", data=df)
plt.title("Tenure Distribution by Churn")
plt.xlabel("Churn Status")
plt.ylabel("Tenure (Months)")
plt.show()
```

**Observation:** _Customers who churn have lower number of months tenure_
**Evidence:** _the median if of 38 which means 38 months for tenure but the plot shows hights churn is among 0-10 months tenure_
**Possible Explanation:** _New customers are actively testing the service during their first year. If they encounter onboarding issues, unexpected costs, or superior competitor offers early on, they cancel quickly. Once a customer passes the initial year mark, brand habituation, trust, and switching friction keep them retained long-term_

```python
# Task 17 — numerical feature 2 vs Churn
charges_stats = df.groupby("Churn")["MonthlyCharges"].describe()
print(charges_stats)

plt.figure(figsize=(7, 4))
sns.boxplot(x="Churn", y="MonthlyCharges", data=df)
plt.title("Monthly Charges Distribution by Churn")
plt.xlabel("Churn Status")
plt.ylabel("Monthly Charges ($)")
plt.show()
```

**Observation:** _Customers with higher monthly bills show a much higher tendency to churn than those paying lower monthly rates_
**Evidence:** _The median monthly charge for churned customers is approximately $79.60, whereas the median for non-churned customers is around $61.25. The distribution for churned users is heavily skewed toward the higher pricing range_
**Possible Explanation:** _High monthly bills increase financial sensitivity and lower customer tolerance for service defects. Subscribers paying premium rates actively monitor their bills and are far more motivated to shop around for cheaper alternative providers._

In Lab 3 you will train a classifier. Before that, you must check whether any column contains information that would not realistically be available at the moment of prediction — using such a column would make a model look artificially accurate.

**Tasks**

**Answer 21 (what would go wrong):**
_If a leaking feature or non-predictive identifier is left in the model's training data, the model will learn artificial, overly optimistic shortcuts that yield near-perfect training accuracy. Once deployed in a real-world production environment to predict churn for active, ongoing customers, the model will fail completely because that future information will not be available at the time of prediction_

## Part 11 — Building the Clean Dataset

Bring together every cleaning decision into one final dataframe, `clean_df`, that Lab 3 can load directly.

> Do **not** split into train/test, encode categorical variables, or train any model — that happens in Lab 3.

**Tasks**
23. Produce `clean_df` with correct data types, missing values handled, duplicates handled, and irrelevant identifier columns removed where justified. Save it (e.g. `clean_churn.csv`) alongside your notebook.
24. Print `df.info()` (or equivalent) for `clean_df` as final proof that every issue identified earlier has been resolved.

```python
# Task 23 — assemble clean_df from your cleaning decisions
clean_df = df.copy()

# 1. Drop irrelevant identifier column
if "customerID" in clean_df.columns:
    clean_df.drop(columns=["customerID"], inplace=True)

# 2. Remove fully duplicated rows
clean_df.drop_duplicates(inplace=True)

# 3. Convert TotalCharges to numeric (converting blank spaces or invalid strings to NaN)
clean_df["TotalCharges"] = pd.to_numeric(clean_df["TotalCharges"], errors="coerce")

# 4. Impute missing values in TotalCharges with the median
total_charges_median = clean_df["TotalCharges"].median()
clean_df["TotalCharges"].fillna(total_charges_median, inplace=True)

# Task 24 — verify clean_df structure and data types
clean_df.info()

# Save the cleaned dataset
clean_df.to_csv("clean_churn.csv", index=False)
```

## Part 12 — Key Findings & Final Reflection

Step back from the code and summarize what you actually learned about this dataset and this problem.

**Task 25 — Five Important Findings**

Repeat this block for each of your five findings.

**Finding 1**
- Observation: _Short-term contracts are heavily associated with customer loss, whereas long-term contracts secure customer retention._
- Evidence: _Month-to-month subscribers churn at approximately 42.7%, while 1-year and 2-year contract subscribers churn at only 11.3% and 2.8% respectively._
- Interpretation: _Customers with month-to-month agreements face no switching barriers, allowing them to cancel immediately upon encountering service friction or better competitor deals._

**Finding 2**
- Observation: _Fiber optic internet users exhibit an unexpectedly high churn rate compared to other connection types._
- Evidence: _Fiber optic customers churn at 41.9%, compared to 19.0% for DSL users and 7.4% for customers with no internet service._
- Interpretation: _Fiber optic service is a premium product. Higher costs create higher customer expectations, making users far more sensitive to price increases or service outages._

**Finding 3**
- Observation: _New customers are at the highest risk of leaving early in their lifecycle._
- Evidence: _The median tenure for churned customers is only 10 months, compared to 38 months for non-churned customers._
- Interpretation: _The first year is critical for customer retention; onboarded users who face unresolved issues during this window leave quickly._

**Finding 4**
- Observation: _Manual, non-automated payment methods lead to higher churn rates than automated billing options._
- Evidence: _Electronic check users churn at 45.3%, whereas customers using automated bank transfers or credit cards churn at around 15–16%._
- Interpretation: _Manual bill payments act as a monthly trigger for customers to re-evaluate their monthly expenses, whereas automated payments create frictionless retention_

**Finding 5**
- Observation: _Cumulative total charges show a strong linear relationship with customer tenure._
- Evidence: _TotalCharges and tenure exhibit a high positive correlation (above 0.82)._
- Interpretation: _TotalCharges is largely a mathematical function of monthly rate multiplied by duration, making it a redundant feature that adds collinearity to linear models._
